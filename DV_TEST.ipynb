```python
average_open = data['Open'].mean()
average_close = data['Close'].mean()
print("Average Open Price:", average_open)
print("Average Close Price:", average_close)
```

```python
highest_price = data['High'].max()
lowest_price = data['Low'].min()
print("Highest Price:", highest_price)
print("Lowest Price:", lowest_price)
```

```python
data["Daily_price_changes"] = data["Close"] - data["Open"]
print(data["Daily_price_changes"])
```

```python
data["Daily_Delta"] = data.head(10)["Close"] - data.head(10)["Open"]
print(data.head(10)["Daily_Delta"])
```

```python
data["Daily_Return"] = ((data["Close"] - data["Open"]) / data["Open"]) * 100
print(data.head(10)["Daily_Return"])
```
